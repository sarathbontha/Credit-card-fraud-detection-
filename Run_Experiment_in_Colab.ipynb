```python
import kagglehub, shutil, pathlib
dataset_dir = pathlib.Path(kagglehub.dataset_download("whenamancodes/fraud-detection"))
matches = list(dataset_dir.rglob("creditcard.csv"))
if not matches:
    raise FileNotFoundError("creditcard.csv was not found in downloaded Kaggle dataset")
shutil.copy2(matches[0], "creditcard.csv")
print("Downloaded:", matches[0])
```
